Loop

```python
fees = [0, 13000, 0, 1000, 18000]
total_fees =0
high_fee_months = 0
for fee in fees:
  if fee!= 0:


    total_fees += fee
  if fee > 12000:
      high_fee_months +=1
print(f"{total_fees} {high_fee_months}")
```
